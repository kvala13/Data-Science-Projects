```python
import json
import requests
from bs4 import BeautifulSoup

def extract_movie_details(movie_url):

    response = requests.get(movie_url, headers=headers)
    soup = BeautifulSoup(response.text, "lxml")

    json_data = None

    for script in soup.find_all("script", type="application/ld+json"):

        try:
            data = json.loads(script.string)

            if isinstance(data, dict) and data.get("@type") == "Movie":
                json_data = data
                break

        except:
            continue

    if json_data is None:
        return None

    # ---------- Director ----------

    director = ""

    if json_data.get("director"):

        director = ", ".join(
            person.get("name","")
            for person in json_data["director"])

    # ---------- Actors ----------

    actors = ""

    if json_data.get("actor"):

        actors = ", ".join(
            actor.get("name","")
            for actor in json_data["actor"])

    # ---------- Rating ----------

    rating_value = ""

    rating_count = ""

    if json_data.get("aggregateRating"):

        rating_value = json_data["aggregateRating"].get("ratingValue","")

        rating_count = json_data["aggregateRating"].get("ratingCount","")

    movie = {

        "Title": json_data.get("name"),

        "Release Date": json_data.get("datePublished"),

        "Description": json_data.get("description"),

        "Genre": ", ".join(json_data.get("genre",[])),

        "Director": director,

        "Actors": actors,

        "Metascore": rating_value,

        "Rating Count": rating_count,

        "URL": movie_url}

    return movie
```

```python
all_movies = []

for link in movies:

    url = "https://www.metacritic.com" + link

    print("Scraping:", url)

    data = extract_movie_details(url)

    if data:

        all_movies.append(data)
```

```python
import requests
from bs4 import BeautifulSoup
import time

BASE_URL = "https://www.metacritic.com"

movie_links = set()

for page in range(1, 101):

    if page == 1:
        url = "https://www.metacritic.com/browse/movie/"
    else:
        url = f"https://www.metacritic.com/browse/movie/?page={page}"

    print(f"Checking Page {page}")

    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print("Page not available")
        continue

    soup = BeautifulSoup(response.text, "lxml")

    links_found = 0

    for a in soup.find_all("a", href=True):

        href = a["href"]

        if href.startswith("/movie/") and href != "/movie/":
            movie_links.add(BASE_URL + href)
            links_found += 1

    print("Movies Found:", links_found)

    time.sleep(1)

print("\nTotal Unique Movies:", len(movie_links))
```

```python
movie_links = list(movie_links)

print(movie_links[:20])

print("Total URLs:", len(movie_links))
```

```python
import pandas as pd
import time

all_movies = []

for i, url in enumerate(movie_links):

    print(f"{i+1}/{len(movie_links)} : {url}")

    try:

        movie = extract_movie_details(url)

        if movie:
            all_movies.append(movie)

    except Exception as e:

        print(e)

    time.sleep(1)
```

```python
df = pd.DataFrame(all_movies)

print(df.shape)

df.head()
```

```python
df.drop_duplicates(subset="Title", inplace=True)

print(df.shape)
```

```python
df.to_csv("metacritic_movies.csv", index=False)

print("Dataset Saved Successfully!")
```
